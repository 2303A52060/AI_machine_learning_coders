Using TensorFlow To Create Text

Using [TensorFlow](https://www.tensorflow.org/?utm_source=chatgpt.com) to create text involves building deep learning models that can understand language patterns and generate human-like sentences. TensorFlow provides powerful tools such as Recurrent Neural Networks (RNNs), Long Short-Term Memory (LSTM) networks, and Transformer models for Natural Language Processing (NLP) tasks. In text generation, the model is first trained on a large dataset of text, where it learns grammar, sentence structure, and word relationships. After training, the model can predict the next word or character in a sequence to generate meaningful text such as stories, poems, chat responses, or code. TensorFlow also includes libraries like Keras, which simplify the process of designing, training, and deploying text generation models efficiently.

```python
from tensorflow.keras.preprocessing.text import Tokenizer

#1.Turning Sequences into input sequences
tokenizer=Tokenizer()
data="In the town of Athy one Jeremy Lanigan \n Battered away ... ..."
corpus=data.lower().split("\n")
tokenizer.fit_on_texts(corpus)
total_words=len(tokenizer.word_index)+1
print(total_words)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

input_sequences=tokenizer.texts_to_sequences(corpus)
max_sequence_len=max([len(x) for x in input_sequences])
input_sequences=np.array(pad_sequences(input_sequences,maxlen=max_sequence_len,padding='pre'))
print(input_sequences)
```

```python
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Create input sequences (xs) and labels (ys)
xs, labels = input_sequences[:,:-1], input_sequences[:,-1]

# One-hot encode the labels
ys = to_categorical(labels, num_classes=total_words)

print("Input sequences (xs) shape:", xs.shape)
print("One-hot encoded labels (ys) shape:", ys.shape)
```

Now that we have our training data (`xs` and `ys`), we can define the neural network model. The model will consist of:

1.  **Embedding Layer**: Converts positive integers (word indices) into dense vectors of fixed size. This layer takes `total_words` as input dimension, `100` as output dimension (embedding size), and `max_sequence_len-1` as input length.
2.  **LSTM Layer**: A type of Recurrent Neural Network (RNN) layer particularly good for sequential data like text. It has `150` units.
3.  **Dense Output Layer**: A fully connected layer with `total_words` units and `softmax` activation, which outputs a probability distribution over the vocabulary for the next word.

```python
model = Sequential()
model.add(Embedding(total_words, 100, input_length=max_sequence_len-1))
model.add(LSTM(150))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

print(model.summary())
```

After defining the model architecture, we compile it with `categorical_crossentropy` as the loss function (suitable for multi-class classification with one-hot encoded labels), `adam` as the optimizer, and `accuracy` as the metric. Now, we'll train the model using our prepared `xs` and `ys` data.

```python
history = model.fit(xs, ys, epochs=100, verbose=1)

print("Model training complete.")
```

```python
send_text="sweet jeremy saw dublin"
next_words=10

for _ in range(next_words):
  token_list=tokenizer.texts_to_sequences([send_text])[0]
  token_list=pad_sequences([token_list],maxlen=max_sequence_len-1,padding='pre')
  predicted_probabilities = model.predict(token_list, verbose=0)
  predicted = np.argmax(predicted_probabilities, axis=-1)
  output_word=""
  for word,index in tokenizer.word_index.items():
    if index==predicted:
      output_word=word
      break
  send_text+=" "+output_word
print(send_text)
```
