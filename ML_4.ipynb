```python
1.#USING TDFS WITH KERAS MODELS
# Import TensorFlow and TensorFlow Datasets
import tensorflow as tf
import tensorflow_datasets as tfds

# Load the Fashion-MNIST dataset (train and test sets)
# tfds.load returns TensorFlow dataset objects, we convert them to NumPy arrays
(training_images, training_labels), (test_images, test_labels) = tfds.as_numpy(
    tfds.load('fashion_mnist',
              split=['train', 'test'],   # split into training and testing sets
              batch_size=-1,             # load all data in one batch
              as_supervised=True)        # return (image, label) pairs
)

# Normalize pixel values (0–255 → 0–1) for better training performance
training_images = training_images / 255.0
test_images = test_images / 255.0

# Build a Sequential neural network model
model = tf.keras.models.Sequential([
    # Flatten layer: converts 28x28 image into a 784-element vector
    tf.keras.layers.Flatten(input_shape=(28,28,1)),

    # Hidden Dense layer: 128 neurons, ReLU activation for non-linearity
    tf.keras.layers.Dense(128, activation=tf.nn.relu),

    # Dropout layer: randomly drops 20% of neurons during training to prevent overfitting
    tf.keras.layers.Dropout(0.2),

    # Output Dense layer: 10 neurons (one per class), softmax activation for probability distribution
    tf.keras.layers.Dense(10, activation=tf.nn.softmax)
])

# Compile the model: define optimizer, loss function, and evaluation metric
model.compile(
    optimizer='adam',                          # Adam optimizer (adaptive learning rate)
    loss='sparse_categorical_crossentropy',    # suitable for integer labels (0–9)
    metrics=['accuracy']                       # track accuracy during training/testing
)

# Train the model on training data for 5 epochs (5 passes through the dataset)
model.fit(training_images, training_labels, epochs=5)
```

```python
2.#USING TDFS WITH KERAS MODELS
import tensorflow as tf
import tensorflow_datasets as tfds

# Load dataset (horses vs humans)
data = tfds.load('horses_or_humans', split='train', as_supervised=True)
train_batches = data.shuffle(100).batch(10)  # shuffle + batch

# Build CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(16, (3,3), activation='relu', input_shape=(300,300,3)), # feature detection
    tf.keras.layers.MaxPooling2D(2,2),  # reduce size

    tf.keras.layers.Conv2D(32, (3,3), activation='relu'), # deeper features
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Conv2D(64, (3,3), activation='relu'), # even deeper features
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Flatten(),  # convert to 1D
    tf.keras.layers.Dense(512, activation='relu'),  # fully connected layer
    tf.keras.layers.Dense(1, activation='sigmoid')  # output: horse or human
])

# Compile model (optimizer + loss + metric)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train model
history = model.fit(train_batches, epochs=10)
```

```python
#Using Mapping Functions for Augmentation
# Import necessary libraries
import tensorflow as tf
import tensorflow_datasets as tfds

# -------------------------------
# 1. Load the dataset
# -------------------------------
# 'horses_or_humans' is a binary classification dataset
# as_supervised=True ensures we get (image, label) pairs
data = tfds.load('horses_or_humans', split='train', as_supervised=True)

# -------------------------------
# 2. Define augmentation function
# -------------------------------
def augment_images(image, label):
    # Convert image to float32 for processing
    image = tf.cast(image, tf.float32)

    # Normalize pixel values to range [0,1]
    image = image / 255.0

    # Apply random horizontal flip
    image = tf.image.random_flip_left_right(image)

    # Apply random brightness adjustment
    image = tf.image.random_brightness(image, max_delta=0.1)

    # Apply random contrast adjustment
    image = tf.image.random_contrast(image, lower=0.9, upper=1.1)

    return image, label

# -------------------------------
# 3. Apply augmentation using map()
# -------------------------------
# The map() function applies augment_images to each (image, label) pair
train = data.map(augment_images)

# -------------------------------
# 4. Shuffle and batch the dataset
# -------------------------------
# Shuffle with buffer size 100, then batch into groups of 32
train_batches = train.shuffle(100).batch(32)

# -------------------------------
# 5. Build a simple CNN model
# -------------------------------
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(300,300,3)),
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(2, activation='softmax)  # 2 classes: horse or human
])

# -------------------------------
# 6. Compile the model
# -------------------------------
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# -------------------------------
# 7. Train the model
# -------------------------------
# Train for a few epochs using augmented batches
history = model.fit(train_batches, epochs=5)

# -------------------------------
# 8. Evaluate (optional)
# -------------------------------
# You can load the test split and evaluate similarly
test_data = tfds.load('horses_or_humans', split='test', as_supervised=True)
test = test_data.map(augment_images).batch(32)
model.evaluate(test)
```
