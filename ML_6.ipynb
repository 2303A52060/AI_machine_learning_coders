```python
#2.
#MAKING SENTIMENT PROGRAMMING USING EMBEDDINGS

#BULDING A SARACASM DETECTOR USING EMBEDDINGS

import tensorflow as tf
model=tf.keras.Sequential([
tf.keras.layers.Embedding(1000,16),
tf.keras.layers.GlobalAveragePooling1D(),
tf.keras.layers.Dense(16,activation='relu'),
tf.keras.layers.Dense(1,activation='sigmoid')
])
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
print(model.summary())
```

To visualize the embeddings, we need to first create a tokenizer to map words to numerical indices and vice versa. Then, we need to extract the weights from the embedding layer of the model. Since the model has not been trained, these embeddings will represent random initializations.

```python
from tensorflow.keras.preprocessing.text import Tokenizer

# Dummy text data to create a tokenizer (replace with your actual text data)
# The vocabulary size should be at least as large as the input_dim of your Embedding layer.
corpus = [
    "This is the first sentence for our vocabulary.",
    "Another sentence to expand the word index.",
    "More words like programming and embeddings are useful here.",
    "sarcasm detector using embeddings is what we are building"
]

# Initialize tokenizer with the same vocab size as the embedding layer's input_dim (1000)
vocab_size = 1000
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<unk>")
tokenizer.fit_on_texts(corpus)

word_index = tokenizer.word_index
# Create a reverse_word_index to map indices back to words
reverse_word_index = dict([(value, key) for (key, value) in word_index.items()])

print(f"Vocabulary size created: {len(word_index)}")
print("First 10 items in word_index:", list(word_index.items())[:10])
print("First 10 items in reverse_word_index:", list(reverse_word_index.items())[:10])
```

```python
import tensorflow as tf

# Extract the weights from the embedding layer
# The embedding layer is the first layer in our model (index 0)

# Ensure the model is built before extracting weights
# The input_shape should match the expected input sequence length of your data.
# From previous cells, we know training_padded has a sequence length of 5.
if not model.built:
    model.build(input_shape=(None, 5)) # (None for batch size, 5 for sequence length)

weights = model.layers[0].get_weights()[0]

print(f"Shape of embedding weights: {weights.shape}")
```

```python
#visulizing the Embeddings

import io

out_v = io.open('vecs.tsv', 'w', encoding='utf-8')
out_m = io.open('meta.tsv', 'w', encoding='utf-8')

# We iterate from 1 because 0 is usually reserved for padding or unknown tokens
# and the word_index starts from 1 with '<unk>'.
# The number of words to iterate over should correspond to the vocab size used in the embedding layer.
for word_num in range(1, vocab_size):
    # Check if the word_num exists in reverse_word_index to handle potential discrepancies
    if word_num in reverse_word_index:
        word = reverse_word_index[word_num]
        embeddings = weights[word_num]
        out_m.write(word + '\n')
        out_v.write('\t'.join([str(x) for x in embeddings]) + '\n')
out_v.close()
out_m.close()

print("Embedding vectors written to vecs.tsv")
print("Embedding metadata written to meta.tsv")
print("You can now use these files with a tool like the TensorFlow Embedding Projector.")
```

```python
#2.using Pretrained Embeddings from TensorFlow Hub
import tensorflow as tf
import tensorflow_hub as hub

# Define a wrapper class to ensure compatibility with tf.keras.Sequential
class HubKerasLayerWrapper(tf.keras.layers.Layer):
    def __init__(self, hub_module_url, output_shape, dtype, trainable=True, **kwargs):
        super(HubKerasLayerWrapper, self).__init__(**kwargs)
        self.hub_layer = hub.KerasLayer(hub_module_url, output_shape=output_shape, dtype=dtype, trainable=trainable)

    def call(self, inputs):
        return self.hub_layer(inputs)

# Instantiate the wrapper layer
hub_layer_wrapped = HubKerasLayerWrapper(
    "https://tfhub.dev/google/tf2-preview/nnlm-en-dim50/1",
    output_shape=[20] , # input_shape=[] argument is handled internally by hub.KerasLayer for scalar string input
    dtype=tf.string, trainable=True
)

model = tf.keras.Sequential([
    tf.keras.Input(shape=(), dtype=tf.string), # Explicitly define the input as a scalar string tensor
    hub_layer_wrapped, # Add the wrapped layer
    tf.keras.layers.Dense(16,activation='relu'),
    tf.keras.layers.Dense(1,activation='sigmoid')
])

adam=tf.keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='binary_crossentropy',optimizer=adam,metrics=['accuracy'])
print(model.summary())
```
