```python
#getting the csv file
import csv
from bs4 import BeautifulSoup # Import BeautifulSoup
from nltk.corpus import stopwords # Import stopwords
import string # Import string for str.maketrans

sentences=[]
labels=[]

# Define table if it's not defined globally or in a previous cell
table = str.maketrans("", "", string.punctuation)

with open('/tmp/binary-emotion.csv',encoding='UTF-8') as csvfile:
  reader = csv.reader(csvfile) # Define reader
  next(reader) # Skip header row if exists, or remove if not
  for row in reader:
    labels.append(int(row[0]))
    # Assuming the sentence is in row[1], adjust if it's different
    current_sentence = row[1] # Assign sentence from csv row

    current_sentence = current_sentence.replace(" , " , ", ") # Fix extra space around comma
    current_sentence = current_sentence.replace(" . " , ". ") # Fix extra space around period
    current_sentence = current_sentence.replace(" ! " , "! ") # Fix extra space around exclamation
    current_sentence = current_sentence.replace(" ? " , "? ") # Fix extra space around question mark

    soup = BeautifulSoup(current_sentence, 'html.parser') # Pass the sentence to BeautifulSoup
    current_sentence = soup.get_text() # Get clean text

    words = current_sentence.lower().split() # Convert to lowercase and split into words
    filtered_sentence_parts = [] # Use a list to build filtered sentence parts

    # Ensure stopwords are loaded (e.g., from nltk.corpus or a predefined set)
    # Assuming stopwords are loaded as 'stopwords' from nltk or a similar source
    # If not, you'll need to define a 'stopwords' set, e.g., stopwords = set(nltk.corpus.stopwords.words('english'))
    stop_words_set = set(stopwords.words('english'))

    for word in words:
      word = word.translate(table) # Apply translation (punctuation removal)
      if word and word not in stop_words_set: # Check if word is not empty after translation
        filtered_sentence_parts.append(word)

    filtered_sentence = " ".join(filtered_sentence_parts) # Join filtered words into a sentence
    sentences.append(filtered_sentence) # Append the fully filtered sentence to the list
```
