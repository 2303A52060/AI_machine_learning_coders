```python
#CHAPTER 5 Introduction to Natural Language Processing
# Import TensorFlow and Keras libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer

# Sample sentences (our dataset)
# In machine learning, text data must be converted into numerical form
sentences = [
    'Today is a sunny day',
    'Today is a rainy day'
]

# Initialize the Tokenizer
# num_words = 100 means we will keep only the top 100 most frequent words
# This is useful when working with large datasets to limit vocabulary size
tokenizer = Tokenizer(num_words=100)

# Fit the tokenizer on our sentences
# This step builds the word index (mapping of words to integers)
tokenizer.fit_on_texts(sentences)

# Retrieve the word index dictionary
# Each unique word is assigned a unique integer ID
word_index = tokenizer.word_index

# Print the word index
# This shows how words are converted into tokens (numbers)
print("Word Index (Vocabulary):")
print(word_index)

# Example: Convert sentences into sequences of tokens
# This is how machine learning models process text data
sequences = tokenizer.texts_to_sequences(sentences)
print("\nTokenized Sequences:")
print(sequences)
```

```python
#Working with Real Data Sources
import tensorflow_datasets as tfds
import tensorflow as tf
from bs4 import BeautifulSoup
import string

# Define stopwords (can extend/customize for domain-specific tasks)
stopwords = ["a", "an", "the", "is", "in", "on", "at", "and", "or", "of", "to", "with", "for", "from", "by", "about", "into", "after", "over", "under", "between", "out", "up", "down", "your", "yours", "yourselves"]

# Translation table for removing punctuation
table = str.maketrans('', '', string.punctuation)

# Container for cleaned sentences
imdb_sentences = []

# Load IMDb dataset (train split)
train_data = tfds.as_numpy(tfds.load('imdb_reviews', split="train"))

for item in train_data:
    # Decode bytes to string and lowercase
    sentence = str(item['text'].decode('UTF-8').lower())

    # Remove HTML tags
    soup = BeautifulSoup(sentence, "html.parser")
    sentence = soup.get_text()

    # Tokenize by splitting
    words = sentence.split()

    # Remove punctuation + stopwords
    filtered_sentence = ""
    for word in words:
        word = word.translate(table)
        if word not in stopwords and word.strip() != "":
            filtered_sentence += word + " "

    imdb_sentences.append(filtered_sentence.strip())

# Tokenizer for converting text to sequences
tokenizer = tf.keras.preprocessing.text.Tokenizer(num_words=25000, oov_token="<OOV>")
tokenizer.fit_on_texts(imdb_sentences)

# Convert sentences to integer sequences
sequences = tokenizer.texts_to_sequences(imdb_sentences)

print("Vocabulary size:", len(tokenizer.word_index))
print("Sample sequence:", sequences[0][:20])  # show first 20 tokens of first review
```

```python
#Getting Text from TensorFlow Datasets
#This refers to the process of loading raw text data from TensorFlow Datasets (TFDS), cleaning it (removing HTML tags, punctuation, and stopwords),
#and preparing it for machine learning models. The goal is to transform unstructured text into numerical sequences that can be used for tasks like
#sentiment analysis, classification, or training neural networks.
import tensorflow_datasets as tfds
import tensorflow as tf
from bs4 import BeautifulSoup
import string

# Define stopwords (shortened for example, you can expand)
stopwords = ["a", "an", "the", "is", "are", "was", "were", "yourselves"]

# Remove punctuation
table = str.maketrans('', '', string.punctuation)

# Load IMDB dataset
train_data = tfds.as_numpy(tfds.load('imdb_reviews', split="train"))

imdb_sentences = []

# Preprocess each review
for item in train_data:
    # Convert bytes to string and lowercase
    sentence = str(item['text'].decode('UTF-8').lower())

    # Remove HTML tags
    soup = BeautifulSoup(sentence, "html.parser")
    sentence = soup.get_text()

    # Tokenize words
    words = sentence.split()
    filtered_sentence = ""

    for word in words:
        # Remove punctuation
        word = word.translate(table)
        # Remove stopwords
        if word not in stopwords and word != "":
            filtered_sentence += word + " "

    imdb_sentences.append(filtered_sentence.strip())

# Tokenizer for converting text to sequences
tokenizer = tf.keras.preprocessing.text.Tokenizer(num_words=25000)
tokenizer.fit_on_texts(imdb_sentences)

# Convert text to sequences
sequences = tokenizer.texts_to_sequences(imdb_sentences)

# Example output
print("Vocabulary size:", len(tokenizer.word_index))
print("First sequence:", sequences[0][:20])  # show first 20 tokens of first review
```

```python
#Getting Text from CSV Files
import csv
from bs4 import BeautifulSoup
import string

# -------------------------------
# 1. Define preprocessing tools
# -------------------------------

# Stopwords list (expand as needed for your project)
stopwords = ["a", "an", "the", "is", "are", "was", "were", "yourselves"]

# Translation table to remove punctuation
table = str.maketrans('', '', string.punctuation)

# Containers for sentences and labels
sentences = []
labels = []

# -------------------------------
# 2. Load and preprocess CSV data
# -------------------------------

# Open the CSV file containing emotion labels and text
with open('/tmp/binary-emotion.csv', encoding='UTF-8') as csvfile:
    reader = csv.reader(csvfile, delimiter=",")

    for row in reader:
        if not row:  # Skip empty rows
            continue

        # First column = label (e.g., 0 = negative, 1 = positive)
        labels.append(int(row[0]))

        # Second column = raw sentence
        sentence = row[1].lower()

        # Add spacing around punctuation for better tokenization
        sentence = sentence.replace(",", " , ")
        sentence = sentence.replace(".", " . ")
        sentence = sentence.replace("-", " - ")
        sentence = sentence.replace("/", " / ")

        # Remove HTML tags if present
        soup = BeautifulSoup(sentence, "html.parser")
        sentence = soup.get_text()

        # Tokenize into words
        words = sentence.split()
        filtered_sentence = ""

        # Remove punctuation + stopwords
        for word in words:
            word = word.translate(table)
            if word not in stopwords and word != "":
                filtered_sentence += word + " "

        # Append cleaned sentence
        sentences.append(filtered_sentence.strip())

# -------------------------------
# 3. Convert text to sequences
# -------------------------------

# Tokenizer for converting text into numerical sequences
tokenizer = tf.keras.preprocessing.text.Tokenizer(num_words=25000)
tokenizer.fit_on_texts(sentences)

# Convert sentences to sequences of integers
sequences = tokenizer.texts_to_sequences(sentences)

# -------------------------------
# 4. Example outputs
# -------------------------------
print("Total samples:", len(sentences))
print("Vocabulary size:", len(tokenizer.word_index))
print("First sequence:", sequences[0][:20])  # show first 20 tokens
print("First label:", labels[0])

#Workflow Explanation
#Load CSV → Each row contains a label and a sentence.

#Clean Text → Lowercase, remove HTML tags, punctuation, and stopwords.

#Tokenize → Split into words.

#Filter → Keep only meaningful words.

#Convert to Sequences → Map words to integers using Tokenizer.

#Ready for ML Models → Sequences + labels can be fed into deep learning models for emotion/sentiment classification.
```

```python
#Topic Definition: Getting Text from JSON Files
#This refers to the process of loading structured text data stored in JSON format, cleaning it (removing punctuation, HTML tags, and stopwords),
#and preparing it for machine learning models. JSON files are widely used because they store data in key‑value pairs, making them easy to parse
#and integrate into ML pipelines. In NLP tasks, JSON often contains text (e.g., headlines, reviews) along with labels (e.g., sarcasm, sentiment).
import json
from bs4 import BeautifulSoup
import string
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import os

# Create a dummy sarcasm.json file if it doesn't exist
json_file_path = '/tmp/sarcasm.json'
if not os.path.exists(json_file_path):
    dummy_json_content = [
        {"headline": "police arrested man for stealing an elephant, he was trying to sell it for peanuts", "is_sarcastic": 1, "article_link": "#"},
        {"headline": "mother-in-law takes vacation to mars, wants to be left alone", "is_sarcastic": 1, "article_link": "#"},
        {"headline": "study finds that children who eat candy are more likely to be happy", "is_sarcastic": 0, "article_link": "#"},
        {"headline": "man buys new car after winning lottery, says he 'deserves' it", "is_sarcastic": 0, "article_link": "#"}
    ]
    with open(json_file_path, 'w') as f:
        json.dump(dummy_json_content, f, indent=4)
    print(f"Dummy JSON file '{json_file_path}' created successfully.")

# -------------------------------
# 1. Define preprocessing tools
# -------------------------------

# Stopwords list (expand as needed)
stopwords = ["a", "an", "the", "is", "are", "was", "were", "yourselves"]

# Translation table to remove punctuation
table = str.maketrans('', '', string.punctuation)

# Containers for sentences, labels, and URLs
sentences = []
labels = []
urls = []

# -------------------------------
# 2. Load and preprocess JSON data
# -------------------------------

# Open sarcasm dataset (JSON format)
with open(json_file_path, 'r') as f:
    datastore = json.load(f)

for item in datastore:
    # Extract headline text and convert to lowercase
    sentence = item['headline'].lower()

    # Add spacing around punctuation for better tokenization
    sentence = sentence.replace(",", " , ")
    sentence = sentence.replace(".", " . ")
    sentence = sentence.replace("-", " - ")
    sentence = sentence.replace("/", " / ")

    # Remove HTML tags if present
    soup = BeautifulSoup(sentence, "html.parser")
    sentence = soup.get_text()

    # Tokenize into words
    words = sentence.split()
    filtered_sentence = ""

    # Remove punctuation + stopwords
    for word in words:
        word = word.translate(table)
        if word not in stopwords and word != "":
            filtered_sentence += word + " "

    # Append cleaned sentence, label, and article link
    sentences.append(filtered_sentence.strip())
    labels.append(item['is_sarcastic'])
    urls.append(item['article_link'])

# -------------------------------
# 3. Train/Test Split
# -------------------------------

training_size = 23000
training_sentences = sentences[0:training_size]
testing_sentences = sentences[training_size:]
training_labels = labels[0:training_size]
testing_labels = labels[training_size:]

# -------------------------------
# 4. Tokenization & Sequence Conversion
# -------------------------------

vocab_size = 20000       # Limit vocabulary size
max_length = 10          # Max sequence length
trunc_type = 'post'      # Truncate sequences at the end
padding_type = 'post'    # Pad sequences at the end
oov_tok = "<OOV>"        # Token for out-of-vocabulary words

# Initialize tokenizer n**Tokenize** → Split into words.n
tokenizer = Tokenizer(num_words=vocab_size, oov_token=oov_tok)
tokenizer.fit_on_texts(training_sentences)

# Convert text to sequences
training_sequences = tokenizer.texts_to_sequences(training_sentences)

# Pad sequences to ensure uniform length
padded = pad_sequences(training_sequences, padding=padding_type, maxlen=max_length)

# -------------------------------
# 5. Example Outputs
# -------------------------------
print("Vocabulary size:", len(tokenizer.word_index))
print("First sequence:", training_sequences[0][:20])
print("First padded sequence:", padded[0])
print("First label:", training_labels[0])


## ⚙️ Workflow Explanation
#1. Load JSON** → Each item contains a headline, sarcasm label, and article link.
#2. **Clean Text** → Lowercase, remove HTML tags, punctuation, and stopwords.
#3. **Tokenize** → Split into words.
#4. **Filter** → Keep only meaningful words.
#5. **Train/Test Split** → Divide dataset into training and testing sets.
#6. **Convert to Sequences** → Map words to integers using `Tokenizer`.
#7. **Pad Sequences** → Ensure all sequences have the same length for ML models.
#8. **Ready for ML Models** → Data can now be fed into deep learning models (e.g., LSTM, GRU, Transformers) for sarcasm detection.
```
